```python
from pyspark.sql import functions as F
from pyspark.sql import types as T

base = "Files/buildmate_raw_data/raw"

spark.sql("""
    CREATE SCHEMA IF NOT EXISTS bronze
""")
```

### billing table

```python
# read raw billing data source

bronze_billing = spark.read.option('header','true').csv(f"{base}/billing/billing_export.csv")

# add two column in this table

bronze_billing = bronze_billing.withColumn('ingested_at',F.current_timestamp()).withColumn('source_file',F.input_file_name())

bronze_billing.show(10)

bronze_billing.count()
```

```python
bronze_billing.printSchema()
```

```python
# save this data into delta table

bronze_billing.write.format('delta')\
.option('overwrite','true')\
.option('mergeSchema','true')\
.saveAsTable('bronze.brz_billing')
```
